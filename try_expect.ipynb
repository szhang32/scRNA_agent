```python
try:
    number = int(input("Enter a number: "))
    print(number)
except ValueError as e:
    # Handle cases where the user inputs letters instead of numbers
    print(f'Type of e: {type(e)}')
    print(f"Invalid input {e}! Please enter a valide integer.")
```

```python
def validate_percentage(value):
    if not (0 <= value <= 100):
        raise ValueError(f"percentage {value} is out of bounds")
    return value

try:
    validate_percentage(150)
except ValueError as e:
    print(f'caught custom error: {e}')
```
